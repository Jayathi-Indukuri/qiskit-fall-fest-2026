```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

simulator = AerSimulator()

print("Libraries imported successfully!")
```

## Part A: Correlated Outcomes

### Prediction Before Execution

Both qubits start in state |00⟩. An H gate creates a superposition on the first qubit, and a CNOT gate entangles the qubits. I predict that outcomes `00` and `11` will dominate, with approximately equal probabilities. The two qubits should produce matching measurement outcomes.

```python
circuit_a = QuantumCircuit(2, 2)

circuit_a.h(0)
circuit_a.cx(0, 1)
circuit_a.measure([0, 1], [0, 1])

print(circuit_a.draw(output="text"))
```

```python
compiled_a = transpile(circuit_a, simulator)

result_a = simulator.run(
    compiled_a,
    shots=100
).result()

counts_a = result_a.get_counts()

print("Part A measurement results:", counts_a)
print("Total shots:", sum(counts_a.values()))
```

```python
fig = plot_histogram(
    counts_a,
    title="Challenge 4 - Part A: Correlated Outcomes"
)
display(fig)
```

## Part B: Opposite Outcomes

### Prediction Before Execution

I will add an X gate to the second qubit of the Bell-state circuit. The X gate flips the second qubit, so I predict that outcomes `01` and `10` will dominate. The two qubits should produce opposite measurement outcomes.

```python
circuit_b = QuantumCircuit(2, 2)

circuit_b.h(0)
circuit_b.cx(0, 1)
circuit_b.x(1)
circuit_b.measure([0, 1], [0, 1])

print(circuit_b.draw(output="text"))
```

```python
compiled_b = transpile(circuit_b, simulator)

result_b = simulator.run(
    compiled_b,
    shots=100
).result()

counts_b = result_b.get_counts()

print("Part B measurement results:", counts_b)
print("Total shots:", sum(counts_b.values()))
```

```python
fig = plot_histogram(
    counts_b,
    title="Challenge 4 - Part B: Opposite Outcomes"
)
display(fig)
```

## Part C: Comparison of Both Circuits

| Feature | Part A | Part B |
|---|---|---|
| Number of qubits | 2 | 2 |
| Gates before measurement | H and CNOT | H, CNOT and X |
| Number of shots | 100 | 100 |
| Expected dominant outcomes | `00` and `11` | `01` and `10` |
| Actual results |{'11': 47, '00': 53} |  {'10': 52, '01': 48}|
| Measurement relationship | Correlated | Opposite outcomes |

## Conclusion

This challenge demonstrated how a small change to a quantum circuit can switch the relationship between two qubits' measurement outcomes. The original Bell-state circuit produces correlated outcomes, while adding an X gate to the second qubit produces opposite outcomes. Both experiments used 100 shots, and their histograms illustrated the change in measurement correlation.
